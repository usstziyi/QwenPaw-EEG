````python
# ============================================================
# 2. 取出一个 Raw 对象
# ============================================================
print("=" * 60)
print("2. 取出单个 Raw 对象")
print("=" * 60)

run = subjects[1]['0train']['0'] # 被试1, 训练session, run0

print(f"类型       : {type(run)}")
print(f"采样率     : {run.info['sfreq']} Hz")
print(f"通道数     : {len(run.ch_names)}")
print(f"通道名     : {run.ch_names}")
print(f"采样点数   : {run.n_times}")
print(f"时长       : {run.n_times / run.info['sfreq']:.1f} s")
print(f"数据形状   : {run.get_data().shape}  (channels × samples)")
print()


````

````python
# ============================================================
# 3. 遍历某个被试的所有 Raw
# ============================================================
print("=" * 60)
print("3. 遍历被试1的所有 Raw") # 所有run
print("=" * 60)

subject_data = subjects[1]

for session_name, runs in subject_data.items():
    print(f"\nSession: {session_name}")
    for run_name, run_obj in runs.items():      
        n_samples = run_obj.n_times
        sfreq = run_obj.info['sfreq']
        duration = n_samples / sfreq
        print(f"  Run {run_name}: {len(run_obj.ch_names)} 通道, "
              f"{n_samples} 采样点, {duration:.1f} s")
print()
````

````python
# ============================================================
# 4. 遍历多个被试的所有 Raw
# ============================================================
print("=" * 60)
print("4. 遍历多个被试")
print("=" * 60)

subjects = dataset.get_data(subjects=[1, 2, 3])

all_raws = []
for subject_id, subject_data in subjects.items():
    for session_name, runs in subject_data.items():
        for run_name, raw_obj in runs.items():
            all_raws.append({
                'subject': subject_id,
                'session': session_name,
                'run': run_name,
                'raw': raw_obj,
            })

df = pd.DataFrame(all_raws)
df.head(36)
````

````python

print(f"总共 {len(all_raws)} 个 Raw 对象\n")
for item in all_raws:
    print(f"  Subject {item['subject']} | "
          f"{item['session']} | "
          f"Run {item['run']} | "
          f"{item['raw'].n_times} samples")

print()

````

````python
# ============================================================
# 5. 对 Raw 做预处理
# ============================================================
print("=" * 60)
print("5. 对 Raw 做预处理")
print("=" * 60)

run_0 = subjects[1]['0train']['0'] # 被试1, 训练session, run0
raw = run_0.copy()

# 带通滤波
raw.filter(l_freq=4.0, h_freq=38.0, fir_design='firwin', verbose=False)
print(f"滤波后: {raw.info['highpass']}~{raw.info['lowpass']} Hz")


# 提取事件通道索引和名称
stim_channel = mne.pick_types(raw.info, eeg=False, eog=False, stim=True, exclude=[])
print(f"事件通道索引: {stim_channel}")
stim_channel_names = [raw.ch_names[chan] for chan in stim_channel]  
print(f"事件通道名称: {stim_channel_names}")

# 提取事件信息
events = mne.find_events(raw,stim_channel=['STI'], shortest_event=0)
print(f"事件数: {len(events)}")
print(f"事件信息形状: {events.shape}") # (事件数, 3): (sample：第几个采样点, previous：, event_id)
print(f"事件ID映射: {dataset.metadata.experiment.events}")
print(f"事件名称映射: {dataset.metadata.experiment.class_labels}")
sfreq = raw.info['sfreq']

#  统计每个事件ID的出现次数
unique_ids, counts = np.unique(events[:, 2], return_counts=True)
print("实际的事件ID:")
for uid, cnt in zip(unique_ids, counts):
    print(f"  ID = {uid}, 出现 {cnt} 次")

event_id = {
    'left_hand':  1,
    'right_hand': 2,
    'foot':       3,
    'tongue':     4,
}


events_df = pd.DataFrame(events, columns=['sample', 'previous', 'event_id'])
events_df['event_name'] = events_df['event_id'].map(event_id)

events_df['time_s'] = events_df['sample'] / sfreq  # 事件发生在第几秒
events_df

````

````python
# ============================================================
# 6. 从 Raw 切 Epoch
# ============================================================
print("=" * 60)
print("6. 从 Raw 切 Epoch")
print("=" * 60)



epochs = mne.Epochs(
    raw,
    events=events,
    event_id=event_id,
    tmin=1.0,
    tmax=4.0,
    baseline=None,
    preload=True,
    verbose=False,
)

X = epochs.get_data()
y = epochs.events[:, 2]

print(f"Epoch shape: {X.shape}  (trials, channels, samples)")
print(f"标签类别   : {np.unique(y)}")
for name, eid in event_id.items():
    print(f"  {name:12s}: {np.sum(y == eid)} trials")
print()

````
