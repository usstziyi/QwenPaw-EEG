```python
"""
BNCI2014_001 — dataset.metadata 接口探索 Demo
"""
import os
import mne
from moabb.datasets import BNCI2014_001

# 数据集
data_dir = os.path.abspath("./../datasets")
mne.set_config("MNE_DATA", data_dir)
dataset = BNCI2014_001()
dataset.download()  # 你已经下载过了，重复调用不会重新下载

dataset = BNCI2014_001()
meta = dataset.metadata
print(type(meta))

# ============================================================
# 1. 一级子对象
# ============================================================
print("=" * 60)
print("1. 一级子对象")
print("=" * 60)

print(f"meta.acquisition       → {type(meta.acquisition).__name__}")
print(f"meta.participants      → {type(meta.participants).__name__}")
print(f"meta.experiment        → {type(meta.experiment).__name__}")
print(f"meta.documentation     → {type(meta.documentation).__name__}")
print(f"meta.preprocessing     → {type(meta.preprocessing).__name__}")
print(f"meta.signal_processing → {type(meta.signal_processing).__name__}")
print(f"meta.data_structure    → {type(meta.data_structure).__name__}")
print(f"meta.paradigm_specific → {type(meta.paradigm_specific).__name__}")
print(f"meta.cross_validation  → {type(meta.cross_validation).__name__}")
print(f"meta.bci_application   → {type(meta.bci_application).__name__}")
print(f"meta.performance       → {type(meta.performance).__name__}")
print(f"meta.tags              → {type(meta.tags).__name__}")
print()

# 一级标量字段
print(f"meta.sessions_per_subject : {meta.sessions_per_subject}")
print(f"meta.runs_per_session     : {meta.runs_per_session}")
print(f"meta.file_format          : {meta.file_format}")
print()
```

```python

# ============================================================
# 2. acquisition — 采集设备信息
# ============================================================
print("=" * 60)
print("2. acquisition — 采集设备信息")
print("=" * 60)

a = meta.acquisition

print(f"sampling_rate            : {a.sampling_rate} Hz")
print(f"n_channels               : {a.n_channels}")
print(f"channel_types            : {a.channel_types}")
print(f"sensors                  : {a.sensors}")
print(f"sensor_type              : {a.sensor_type}")
print(f"reference                : {a.reference}")
print(f"ground                   : {a.ground}")
print(f"hardware                 : {a.hardware}")
print(f"software                 : {a.software}")
print(f"filters                  : {a.filters}")
print(f"line_freq                : {a.line_freq} Hz")
print(f"montage                  : {a.montage}")
print(f"cap_manufacturer         : {a.cap_manufacturer}")
print(f"cap_model                : {a.cap_model}")
print(f"electrode_type           : {a.electrode_type}")
print(f"electrode_material       : {a.electrode_material}")
print(f"impedance_threshold_kohm : {a.impedance_threshold_kohm}")
print()

# 嵌套对象: auxiliary_channels
aux = a.auxiliary_channels
print("auxiliary_channels:")
print(f"  has_eog        : {aux.has_eog}")
print(f"  eog_channels   : {aux.eog_channels}")
print(f"  eog_type       : {aux.eog_type}")
print(f"  has_emg        : {aux.has_emg}")
print(f"  emg_channels   : {aux.emg_channels}")
print()
```

```python

# ============================================================
# 3. participants — 被试信息
# ============================================================
print("=" * 60)
print("3. participants — 被试信息")
print("=" * 60)

p = meta.participants

print(f"n_subjects         : {p.n_subjects}")
print(f"health_status      : {p.health_status}")
print(f"species            : {p.species}")
print(f"gender             : {p.gender}")
print(f"age_mean           : {p.age_mean}")
print(f"age_std            : {p.age_std}")
print(f"age_min            : {p.age_min}")
print(f"age_max            : {p.age_max}")
print(f"ages               : {p.ages}")
print(f"handedness         : {p.handedness}")
print(f"sexes              : {p.sexes}")
print(f"handedness_list    : {p.handedness_list}")
print(f"clinical_population: {p.clinical_population}")
print(f"bci_experience     : {p.bci_experience}")
print()
```
